```python
import os
import requests
import netCDF4
import numpy as np
from datetime import datetime, timedelta, timezone
from astral import LocationInfo
from astral.sun import sun
from pathlib import Path

# Float and location info
float_id = "7901133"
latitude = 36.249
longitude = -8.560
start_date = datetime(2024, 6, 1, tzinfo=timezone.utc)
data_dir = Path("profiles")
data_dir.mkdir(exist_ok=True)

# --- Step 1: Download profile files if needed ---
base_url = f"https://data-argo.ifremer.fr/dac/coriolis/{float_id}/profiles/"
max_profiles = 100  # Adjust as needed

for i in range(max_profiles):
    fname = f"R{float_id}_{i:03d}.nc"
    fpath = data_dir / fname
    if not fpath.exists():
        print(f"Downloading {fname}...")
        url = base_url + fname
        r = requests.get(url)
        if r.status_code == 200:
            fpath.write_bytes(r.content)
        else:
            break  # Stop when we run out of files

# --- Step 2: Process each profile file ---
def is_daytime(utc_time, lat, lon):
    loc = LocationInfo(latitude=lat, longitude=lon)
    s = sun(loc.observer, date=utc_time.date(), tzinfo=timezone.utc)
    return s['sunrise'] <= utc_time <= s['sunset']

for nc_file in sorted(data_dir.glob("R*.nc")):
    with netCDF4.Dataset(nc_file) as ds:
        try:
            # Get time (UTC)
            juld = ds.variables["JULD"][:].data[0]
            juld_units = ds.variables["JULD"].units
            time_utc = netCDF4.num2date(juld, juld_units)
        except Exception as e:
            print(f"Skipping {nc_file.name} (invalid time): {e}")
            continue

        if time_utc < start_date:
            continue

        # Daytime check
        if not is_daytime(time_utc, latitude, longitude):
            continue

        # Surface QC check
        try:
            pres_qc = ds.variables["PRES_QC"][:, 0]  # first level
            if isinstance(pres_qc[0], bytes):
                pres_qc = [qc.decode() for qc in pres_qc]
            if pres_qc[0] != '1':
                continue
        except Exception as e:
            print(f"Skipping {nc_file.name} (missing QC): {e}")
            continue

        # Match found
        print("✅ MATCH FOUND:")
        print(f"File: {nc_file.name}")
        print(f"UTC Time: {time_utc}")
        local_time = time_utc + timedelta(minutes=int(longitude * 4))
        print(f"Approx. Local Time: {local_time}")
        break
else:
    print("No matching profile found.")
```
