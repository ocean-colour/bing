# Argo BGC $b_{bp}$ data

```python
# imports
import os
import numpy as np

import pandas

from remote_sensing.plotting import globe
```

# Load me

```python
df = pandas.read_csv('argo_bgc_profiles_bbp.csv')
df.head()
```

# Cut on date

```python
time = pandas.to_datetime(df.time)
```

```python
old_enough = time > pandas.Timestamp('2024-04-01')
np.sum(old_enough)
```

```python
argo_pace = df[old_enough]
```

# Explore a bit

## Geographical plots

## All

```python
globe.plot_lons_lats_vals(df.lon, df.lat, np.ones_like(df.lon), add_colorbar=False)
```

## Old enough for PACE

```python
globe.plot_lons_lats_vals(argo_pace.lon, argo_pace.lat, argo_pace.solar_angle.values,
                         cmap='seismic', cb_lbl='Solar angle (deg)')
```
